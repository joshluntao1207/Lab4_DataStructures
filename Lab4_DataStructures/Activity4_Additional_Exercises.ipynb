```python
# Student Parameters
LAST_NAME = "LUNTAO"
STUDENT_ID = "TUPM-26-1947"
SEED_DIGIT = int(STUDENT_ID[-1])  # 7
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 29
NAME_LENGTH = len(LAST_NAME)  # 6

# Data Generation Rules
# Student IDs generated using SEED_DIGIT, ID_SUM, and NAME_LENGTH
students = [
    (f"STU-{100 + SEED_DIGIT * i}", f"Student_{chr(65+i)}", 75 + ((ID_SUM * (i + 1) + NAME_LENGTH) % 25))
    for i in range(5)
]

# Analysis & Summarization
total_score = sum(score for _, _, score in students)
average_score = total_score / len(students)
highest_student = max(students, key=lambda s: s[2])
lowest_student = min(students, key=lambda s: s[2])

print("=== EXERCISE 1: STUDENT PERFORMANCE RECORDS ===")
print("Generated Student Records (Tuples):")
for sid, name, score in students:
    print(f"  ID: {sid} | Name: {name} | Score: {score}")

print("\nAssessment Data / Results:")
print(f"  Total Class Score: {total_score}")
print(f"  Average Class Score: {average_score:.2f}")
print(f"  Highest Performing Student: {highest_student[1]} ({highest_student[0]}) with Score {highest_student[2]}")
print(f"  Lowest Performing Student: {lowest_student[1]} ({lowest_student[0]}) with Score {lowest_student[2]}")
```

```python
from collections import defaultdict

FAVORITE_ARTIST = "RAKIM"
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 5
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())  # 52

# Data Generation Rules
# Generating transaction records: (Transaction_ID, Customer_ID, Amount)
raw_transactions = [
    (f"TXN-{1000 + i}", f"CUST-{10 + (i % 3)}", (ARTIST_SUM * (i + 1) + SEED_DIGIT * 5) % 200 + 50)
    for i in range(8)
]

# Aggregation using defaultdict
customer_totals = defaultdict(float)
customer_counts = defaultdict(int)

for txn_id, cust_id, amount in raw_transactions:
    customer_totals[cust_id] += amount
    customer_counts[cust_id] += 1

highest_spender = max(customer_totals.items(), key=lambda x: x[1])

print("=== EXERCISE 2: TRANSACTION HISTORY ANALYZER ===")
print("Raw Transactions:")
for t_id, c_id, amt in raw_transactions:
    print(f"  {t_id} | Customer: {c_id} | Amount: ${amt:.2f}")

print("\nCustomer Summaries (Aggregated):")
for cust_id in customer_totals:
    print(f"  Customer: {cust_id} | Total Spent: ${customer_totals[cust_id]:.2f} | Total Txns: {customer_counts[cust_id]}")

print("\nAssessment Data / Results:")
print(f"  Top Spender: {highest_spender[0]} with total expenditure of ${highest_spender[1]:.2f}")
```
